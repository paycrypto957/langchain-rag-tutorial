```python
from langchain_community.document_loaders import TextLoader

text_documents = TextLoader("../knowledge_base/sample.txt", encoding="utf-8").load()
print(text_documents)
```

```python
from langchain_community.document_loaders import UnstructuredMarkdownLoader
md_documents = UnstructuredMarkdownLoader("../knowledge_base/sample.md", mode="elements", strategy="fast").load()
print(md_documents)
```

```python
from langchain_community.document_loaders import UnstructuredPDFLoader

# 测试 OCR PDF
loader = UnstructuredPDFLoader(
    "../knowledge_base/sample.pdf",
    mode="elements",
    strategy="hi_res",
    infer_table_structure=True,
    languages=["chi_sim", "eng"]
)
docs = loader.load()
print(docs)
##print(docs[0].page_content[:300])
```

```python
from langchain_community.document_loaders import UnstructuredWordDocumentLoader
word_doc = UnstructuredWordDocumentLoader("../knowledge_base/sample.docx", mode="elements", strategy="fast").load()
print(word_doc)
```
