### 检索Chroma

```python
from contourpy.util.data import simple
from langchain_chroma import Chroma
import torch
from langchain_huggingface import HuggingFaceEmbeddings
#加载嵌入模型
embedding_model =HuggingFaceEmbeddings(
model_name="../bge-base-zh-v1.5",
model_kwargs={"device": "cuda" if torch.cuda.is_available() else "cpu"},
encode_kwargs={"normalize_embeddings": True}, #输出归一化向量，更适合余弦相似度计算
)
vectorstore = Chroma(embedding_function=embedding_model,persist_directory="vectorstore",)

query = "不动产或者动产被占有人占有怎么办"
simple_doc = vectorstore.similarity_search(query,k=3)
for doc in simple_doc:
    print(doc)
```

```python
query = "不动产或者动产被占有人占有怎么办"
#获取向量存储检索器
retriever = vectorstore.as_retriever(
search_type="similarity",#检索方式，similarity或mmr
search_kwargs={"k": 3},
)
sim_docs = retriever.invoke(query)
for doc in sim_docs:
    print(doc)
```
